```python
import re
s=input("Enter a string:")
pattern=r"^[a-zA-Z0-9]+$"
if re.fullmatch(pattern,s):
    print("String contains only a-z,A-Z and 0-9")
else:
    print("String contains other characters also")
```

```python
#slicing in strings
s="Python"
print(s[0])
print(s[-1])
print(s[1:4])
print(s[:3])
print(s[2:])
print(s[::-1])
```

```python
#update
s="Python"
s="J"+s[1:]
s=s.replace("J","P")
s=s+"3"
print(s)
```

```python
#delete

s="Python"
s=s[:2]+s[3:]
print(s)
```
